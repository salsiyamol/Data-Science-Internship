```python
#string is palindrome
string=input("Enter a word")
if string==string[::-1]:
    print("It is a palindrome")
else:
    print("It is not a palindrome")
```

```python
#count vowels
string=input("Enter a String")
vowels="aeiouAEIOU"
v=0
c=0
for ch in string:
    if ch.isalpha():
        if ch in vowels:
            v+=1
        else:
            c+=1
print("vowels:",v)
print("consonants:",c)
```

```python
#frequency of each character
text="hello world"
freq={}
for char in text:
    if char in freq:
        freq[char]+=1
    else:
        freq[char]=1
print(freq)
```
