```python
import random
words = ["python", "apple", "grape", "mango", "banana"]
word = random.choice(words)
stages = [
    """
     -----
     |   |
         |
         |
         |
         |
    --------
    """,
    """
     -----
     |   |
     O   |
         |
         |
         |
    --------
    """,
    """
     -----
     |   |
     O   |
     |   |
         |
         |
    --------
    """,
    """
     -----
     |   |
     O   |
    /|   |
         |
         |
    --------
    """,
    """
     -----
     |   |
     O   |
    /|\\  |
         |
         |
    --------
    """,
    """
     -----
     |   |
     O   |
    /|\\  |
    /    |
         |
    --------
    """,
    """
     -----
     |   |
     O   |
    /|\\  |
    / \\  |
         |
    --------
    """
]
guessed_letters = []
wrong_attempts = 0
max_attempts = 6
display = ["_"] * len(word)
print("Welcome to Hangman!")
while wrong_attempts < max_attempts and "_" in display:
    print(stages[wrong_attempts])
    print("Word:", " ".join(display))
    print("Guessed letters:", guessed_letters)
    guess = input("Enter a letter: ").lower()
    if guess in guessed_letters:
        print("Already guessed!")
        continue
    guessed_letters.append(guess)
    if guess in word:
        for i in range(len(word)):
            if word[i] == guess:
                display[i] = guess
        print("Correct guess!")
    else:
        wrong_attempts += 1
        print(f"Wrong guess! Attempts left: {max_attempts - wrong_attempts}")
print(stages[wrong_attempts])
if "_" not in display:
    print("You won! The word is:", word)
else:
    print("Game Over! The word was:", word)
```
