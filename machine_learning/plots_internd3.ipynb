```python
import matplotlib.pyplot as plt
x=[1, 2, 3, 4, 5]
y=[2, 4, 6, 8, 10]
plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

```python
import matplotlib.pyplot as plt
x=["A", "B", "C", "D"]
y=[10, 20, 15, 25]
plt.bar(x, y)
plt.title("Bar Plot")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1, 2, 3, 4, 5]
y=[5, 7, 8, 6, 9]
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

```python
import matplotlib.pyplot as plt
data=[30, 20, 25, 25]
labels=["Python", "Java", "C", "C++"]
plt.pie(data, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt
data=[10, 20, 20, 30, 40, 40, 40, 50]
plt.hist(data)
plt.title("Histogram")
plt.show()
```

```python
import matplotlib.pyplot as plt
data=[10, 20, 30, 40, 50]
plt.boxplot(data)
plt.title("Box Plot")
plt.show()
```
